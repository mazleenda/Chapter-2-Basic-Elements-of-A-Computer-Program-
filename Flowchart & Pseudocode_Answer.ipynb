**Exercise 2.1 – Flowchart**

Question - A programmer has been assigned to develop a simple roller coaster entrance system.

The system should check whether a user's height meets the minimum safety requirement. If the user's height is less than 130 cm, the system should display a warning message.

**Task**

Draw a flowchart to solve the above problem.

**Submission:**

Draw the flowchart on paper and upload a photo, or
Use an online flowchart tool (e.g., draw.io/diagrams.net) and upload the completed flowchart.

```python
# START
#   ↓
# INPUT height
#   ↓
# Is height < 130 cm?
#   ├── YES → DISPLAY "Warning: Height does not meet the safety requirement"
#   └── NO  → DISPLAY "You are allowed to enter"
#   ↓
# END
```

**Exercise 2.4 – Pseudocode**

Question - Write a pseudocode to calculate the area of a rectangle.

The program should:

Read the length.
Read the width.
Calculate the area.
Display the area.
Task

Write the complete pseudocode.

```python
# BEGIN
#     INPUT length
#     INPUT width
#
#     SET area ← length × width
#
#     DISPLAY area
# END
```

**Exercise 2.6  –Pseudocode**
Question- Write a pseudocode to calculate the Body Mass Index (BMI).

The formula is:

BMI = Weight (kg) / Height² (m²)

Task

Write the complete pseudocode.

```python
# BEGIN
#     INPUT weight
#     INPUT height
#
#     SET bmi ← weight / (height × height)
#
#     DISPLAY bmi
# END
```
